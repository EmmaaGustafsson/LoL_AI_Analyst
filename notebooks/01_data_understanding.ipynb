## Summary

The dataset includes 242 572 observations from 24 912 unique matches.

Every match is represented by multiple time-stamps (frames), which starts from 10 minutes and continues forward into later stages of the game.

The target variable hasWon is balanced (0 = 50.2 %, 1 = 49.8 %).

No missing values or duplicates were identified.

The dataset seems to include accumulated game statistics from each time-stamp of the game.
